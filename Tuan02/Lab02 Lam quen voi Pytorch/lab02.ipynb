## Bài 9


```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X),
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

## Bài 10


```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1 * y1 + y2 * y2 + y3 * y3 + y4 * y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
import torch

X = torch.arange(24)
print("X reshape 4 x 6:\n", X.reshape(4, 6))
print("X reshape 2 x 12:\n", X.reshape(2, 12))
print("X reshape 2 x 3 x 4:\n", X.reshape(2, 3, 4))

A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Mean by column:\n", A.mean(dim=0))
print("Standard deviation by column:\n", A.std(dim=0))

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x ** 2 + 5 * x + 3
y.backward()
print("dy/dx at x=2:", x.grad)
print("Manual result:", 4 * 2 + 5)

torch.manual_seed(0)
sample_x = torch.randn(20, 1) * 5
sample_y = 2 * sample_x ** 2 - 3 * sample_x + 5
features = torch.cat([sample_x ** 2, sample_x, torch.ones_like(sample_x)], dim=1)
weights = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([weights], lr=0.01)

for _ in range(1000):
    optimizer.zero_grad()
    prediction = features @ weights
    loss = torch.mean((sample_y - prediction) ** 2)
    loss.backward()
    optimizer.step()

print("Learned coefficients [x^2, x, 1]:")
print(weights)
```
